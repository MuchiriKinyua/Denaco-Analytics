```python
import random
import numpy as np
import pandas as pd
from datetime import datetime
```

```python
# Load your original file
df = pd.read_excel('Original Sales data.xlsx')
df
```

```python
# Add ~5% missing values
n_missing = int(df.size * 0.05)
for _ in range(n_missing):
    i = random.randint(0, df.shape[0] - 1)
    j = random.randint(0, df.shape[1] - 1)
    df.iat[i, j] = np.nan
```

```python
# Add ~3% duplicate rows
n_duplicates = int(df.shape[0] * 0.03)
duplicates = df.sample(n=n_duplicates, random_state=42)
df = pd.concat([df, duplicates], ignore_index=True)
```

```python
# Mix date formats in 'Order Date' and 'Ship Date'
def random_date_format(date):
    formats = [
        "%Y-%m-%d",        # 2017-11-08
        "%m/%d/%Y",        # 11/08/2017
        "%d-%m-%Y",        # 08-11-2017
        "%b %d, %Y",       # Nov 8, 2017
        "%d %b %Y",        # 8 Nov 2017
    ]
    fmt = random.choice(formats)
    try:
        return pd.to_datetime(date).strftime(fmt)
    except:
        return date

for col in ['Order Date', 'Ship Date']:
    mask = df[col].notnull()
    idx = df[mask].sample(frac=0.2, random_state=42).index  # 20% of dates
    df.loc[idx, col] = df.loc[idx, col].apply(random_date_format)
```

```python
# Save the modified file
df.to_excel('Modified-Sales-data.xlsx', index=False)
print("Done! Your file is saved as 'Modified-Sales-data.xlsx'.")
```
